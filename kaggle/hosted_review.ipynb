## Research question and inputs

How much evidence is lost when sequential strategy validation ignores shared execution-cost structure, or treats a stationary cost coefficient as though it could change adversarially each period? The preserved study contains 24 baseline and 25 separately specified follow-up synthetic cases, with 4,000 independent worlds within each case and an 800-step horizon. Streams and some factorial cells are dependent. The retrospective market study has six cases, nine portfolios and five assumed cost levels, plus a separately dated exploratory daily-bot comparison. Raw and normalized vendor candles are not distributed in this companion.

## Method and key recorded findings

The synthetic comparisons use established bounded betting and empirical-Bernstein constructions, shared versus separate cost bounds, adaptive stakes and whole-history nuisance profiling. Fixed mixtures and adaptive betting have different null scopes. The market books use prior-close signals, next-open self-financing accounting, inherited holdings and modeled traded-notional costs.

At effect 0.05 with equal turnover in the baseline bounded experiment, joint and separate first-stream powers were 92.425% and 0.025%. That controlled finding is not a market edge. At 25 bps per traded side, the primary January 2025–October 2026 fixed and adaptive four-hour EMA paths both returned 2.6067%, with no evaluation-window switches. Hourly/daily adaptive differences were negative and imprecise. Both central matched-settings daily-bot contrast intervals include zero. No new general theorem or demonstrated adaptive trading edge is claimed.

## Limits

Bounded-model sequential guarantees require the stated information and calibration assumptions; they are not transferred to OHLCV returns. Market costs are scenarios, not calibrated live fills. The same Bitcoin history underlies cross-venue checks, bootstrap intervals are approximate and unadjusted, and the later daily-bot contrast is post-outcome exploratory. No capacity, latency, authenticated-fill or prospective-profitability result is established. This notebook inspects saved records and does not rerun the original experiments.
